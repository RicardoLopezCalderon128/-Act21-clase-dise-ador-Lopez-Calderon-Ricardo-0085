#### Diccionario de chef

```python
# Diccionario principal
Diccionario_chef = []

# Función para agregar un chef
def agregar_chef():
    chef = {}
    chef['Nombre'] = input("Ingrese el nombre del chef: ")
    chef['Estrellas Michelin'] = input("Ingrese las estrellas Michelin: ")
    chef['Especialidad Culinaria'] = input("Ingrese la especialidad culinaria: ")
    chef['Restaurante'] = input("Ingrese el nombre del restaurante: ")
    chef['Brigada a Cargo'] = input("Ingrese la brigada a cargo: ")
    chef['Menú Insignia'] = input("Ingrese el menú insignia: ")
    chef['Años en Cocina'] = input("Ingrese los años de experiencia en cocina: ")
    Diccionario_chef.append(chef)
    print("\nChef agregado con éxito.\n")
    listar_chef()

# Función para listar todos los chefs
def listar_chef():
    if len(Diccionario_chef) == 0:
        print("No hay chefs registrados.\n")
        return
    print("\nListado de Chefs:")
    print("{:<20} {:<20} {:<25} {:<20} {:<20} {:<20} {:<10}".format(
        "Nombre", "Estrellas", "Especialidad", "Restaurante", "Brigada", "Menú Insignia", "Años"))
    print("-"*135)
    for chef in Diccionario_chef:
        print("{:<20} {:<20} {:<25} {:<20} {:<20} {:<20} {:<10}".format(
            chef['Nombre'], chef['Estrellas Michelin'], chef['Especialidad Culinaria'],
            chef['Restaurante'], chef['Brigada a Cargo'], chef['Menú Insignia'], chef['Años en Cocina']))
    print()

# Función para buscar un chef por nombre
def buscar_chef():
    nombre = input("Ingrese el nombre del chef a buscar: ")
    encontrado = False
    for chef in Diccionario_chef:
        if chef['Nombre'].lower() == nombre.lower():
            print("\nChef encontrado:")
            print("{:<20} {:<20} {:<25} {:<20} {:<20} {:<20} {:<10}".format(
                "Nombre", "Estrellas", "Especialidad", "Restaurante", "Brigada", "Menú Insignia", "Años"))
            print("-"*135)
            print("{:<20} {:<20} {:<25} {:<20} {:<20} {:<20} {:<10}".format(
                chef['Nombre'], chef['Estrellas Michelin'], chef['Especialidad Culinaria'],
                chef['Restaurante'], chef['Brigada a Cargo'], chef['Menú Insignia'], chef['Años en Cocina']))
            encontrado = True
    if not encontrado:
        print("Chef no encontrado.\n")
    print()
    listar_chef()

# Función para borrar un chef por nombre
def borrar_chef():
    nombre = input("Ingrese el nombre del chef a borrar: ")
    for i, chef in enumerate(Diccionario_chef):
        if chef['Nombre'].lower() == nombre.lower():
            Diccionario_chef.pop(i)
            print("Chef borrado con éxito.\n")
            listar_chef()
            return
    print("Chef no encontrado.\n")
    listar_chef()

# Función para actualizar un chef por nombre
def actualizar_chef():
    nombre = input("Ingrese el nombre del chef a actualizar: ")
    for chef in Diccionario_chef:
        if chef['Nombre'].lower() == nombre.lower():
            print("Ingrese los nuevos datos del chef (dejar en blanco para no cambiar):")
            nuevo_nombre = input("Nombre [{}]: ".format(chef['Nombre']))
            if nuevo_nombre != "":
                chef['Nombre'] = nuevo_nombre
            estrellas = input("Estrellas Michelin [{}]: ".format(chef['Estrellas Michelin']))
            if estrellas != "":
                chef['Estrellas Michelin'] = estrellas
            especialidad = input("Especialidad Culinaria [{}]: ".format(chef['Especialidad Culinaria']))
            if especialidad != "":
                chef['Especialidad Culinaria'] = especialidad
            restaurante = input("Restaurante [{}]: ".format(chef['Restaurante']))
            if restaurante != "":
                chef['Restaurante'] = restaurante
            brigada = input("Brigada a Cargo [{}]: ".format(chef['Brigada a Cargo']))
            if brigada != "":
                chef['Brigada a Cargo'] = brigada
            menu = input("Menú Insignia [{}]: ".format(chef['Menú Insignia']))
            if menu != "":
                chef['Menú Insignia'] = menu
            años = input("Años en Cocina [{}]: ".format(chef['Años en Cocina']))
            if años != "":
                chef['Años en Cocina'] = años
            print("Chef actualizado con éxito.\n")
            listar_chef()
            return
    print("Chef no encontrado.\n")
    listar_chef()

# Menú interactivo
while True:
    print("Menú:")
    print("1. Agregar chef")
    print("2. Listar chefs")
    print("3. Buscar chef")
    print("4. Borrar chef")
    print("5. Actualizar chef")
    print("6. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        agregar_chef()
    elif opcion == "2":
        listar_chef()
    elif opcion == "3":
        buscar_chef()
    elif opcion == "4":
        borrar_chef()
    elif opcion == "5":
        actualizar_chef()
    elif opcion == "6":
        print("Saliendo del programa...")
        break
    else:
        print("Opción inválida, intente nuevamente.\n")
```
