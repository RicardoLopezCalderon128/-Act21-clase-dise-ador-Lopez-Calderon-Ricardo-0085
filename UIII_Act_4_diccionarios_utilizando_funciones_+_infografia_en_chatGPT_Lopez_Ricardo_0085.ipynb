#### Diccionario de chef

```python
# ==============================
# Diccionario principal
# ==============================
Diccionario_chef = {}

# ==============================
# Función para agregar un chef
# ==============================
def agregar_chef():
    chef = {}

    chef['Nombre'] = input("Ingrese el nombre del chef: ")
    chef['Estrellas Michelin'] = input("Ingrese las estrellas Michelin: ")
    chef['Especialidad Culinaria'] = input("Ingrese la especialidad culinaria: ")
    chef['Restaurante'] = input("Ingrese el nombre del restaurante: ")
    chef['Brigada a Cargo'] = input("Ingrese la brigada a cargo: ")
    chef['Menú Insignia'] = input("Ingrese el menú insignia: ")
    chef['Años en Cocina'] = input("Ingrese los años de experiencia en cocina: ")

    Diccionario_chef.append(chef)

    print("\nChef agregado con éxito.\n")
    listar_chef()


# ==============================
# Función para mostrar tabla
# ==============================
def mostrar_encabezado():
    print("=" * 170)

    print("{:<20} {:<22} {:<28} {:<22} {:<22} {:<28} {:<15}".format(
        "NOMBRE",
        "ESTRELLAS",
        "ESPECIALIDAD",
        "RESTAURANTE",
        "BRIGADA",
        "MENU INSIGNIA",
        "AÑOS"
    ))

    print("=" * 170)


# ==============================
# Función para listar chefs
# ==============================
def listar_chef():

    if len(Diccionario_chef) == 0:
        print("No hay chefs registrados.\n")
        return

    print("\nLISTADO DE CHEFS\n")

    mostrar_encabezado()

    for chef in Diccionario_chef:

        print("{:<20} {:<22} {:<28} {:<22} {:<22} {:<28} {:<15}".format(
            chef['Nombre'],
            chef['Estrellas Michelin'],
            chef['Especialidad Culinaria'],
            chef['Restaurante'],
            chef['Brigada a Cargo'],
            chef['Menú Insignia'],
            chef['Años en Cocina']
        ))

    print("=" * 170)
    print()


# ==============================
# Función para buscar chef
# ==============================
def buscar_chef():

    nombre = input("Ingrese el nombre del chef a buscar: ")

    encontrado = False

    for chef in Diccionario_chef:

        if chef['Nombre'].lower() == nombre.lower():

            print("\nCHEF ENCONTRADO\n")

            mostrar_encabezado()

            print("{:<20} {:<22} {:<28} {:<22} {:<22} {:<28} {:<15}".format(
                chef['Nombre'],
                chef['Estrellas Michelin'],
                chef['Especialidad Culinaria'],
                chef['Restaurante'],
                chef['Brigada a Cargo'],
                chef['Menú Insignia'],
                chef['Años en Cocina']
            ))

            print("=" * 170)
            print()

            encontrado = True

    if not encontrado:
        print("\nChef no encontrado.\n")


# ==============================
# Función para borrar chef
# ==============================
def borrar_chef():

    nombre = input("Ingrese el nombre del chef a borrar: ")

    for i, chef in enumerate(Diccionario_chef):

        if chef['Nombre'].lower() == nombre.lower():

            Diccionario_chef.pop(i)

            print("\nChef borrado con éxito.\n")

            listar_chef()

            return

    print("\nChef no encontrado.\n")


# ==============================
# Función para actualizar chef
# ==============================
def actualizar_chef():

    nombre = input("Ingrese el nombre del chef a actualizar: ")

    for chef in Diccionario_chef:

        if chef['Nombre'].lower() == nombre.lower():

            print("\nIngrese los nuevos datos del chef")
            print("Deje en blanco para no cambiar.\n")

            nuevo_nombre = input(f"Nombre [{chef['Nombre']}]: ")
            if nuevo_nombre != "":
                chef['Nombre'] = nuevo_nombre

            estrellas = input(f"Estrellas Michelin [{chef['Estrellas Michelin']}]: ")
            if estrellas != "":
                chef['Estrellas Michelin'] = estrellas

            especialidad = input(f"Especialidad Culinaria [{chef['Especialidad Culinaria']}]: ")
            if especialidad != "":
                chef['Especialidad Culinaria'] = especialidad

            restaurante = input(f"Restaurante [{chef['Restaurante']}]: ")
            if restaurante != "":
                chef['Restaurante'] = restaurante

            brigada = input(f"Brigada a Cargo [{chef['Brigada a Cargo']}]: ")
            if brigada != "":
                chef['Brigada a Cargo'] = brigada

            menu = input(f"Menú Insignia [{chef['Menú Insignia']}]: ")
            if menu != "":
                chef['Menú Insignia'] = menu

            años = input(f"Años en Cocina [{chef['Años en Cocina']}]: ")
            if años != "":
                chef['Años en Cocina'] = años

            print("\nChef actualizado con éxito.\n")

            listar_chef()

            return

    print("\nChef no encontrado.\n")


# ==============================
# Menú principal
# ==============================
while True:

    print("========== MENÚ ==========")
    print("1. Agregar chef")
    print("2. Listar chefs")
    print("3. Buscar chef")
    print("4. Borrar chef")
    print("5. Actualizar chef")
    print("6. Salir")
    print("==========================")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        agregar_chef()

    elif opcion == "2":
        listar_chef()

    elif opcion == "3":
        buscar_chef()

    elif opcion == "4":
        borrar_chef()

    elif opcion == "5":
        actualizar_chef()

    elif opcion == "6":
        print("\nSaliendo del programa...")
        break

    else:
        print("\nOpción inválida, intente nuevamente.\n")1
```
