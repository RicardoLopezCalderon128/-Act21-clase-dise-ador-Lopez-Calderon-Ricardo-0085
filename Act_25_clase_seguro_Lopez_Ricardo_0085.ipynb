Descripción del Proyecto: Clase Seguro
El objetivo es desarrollar un sistema de gestión de pólizas utilizando Programación Orientada a Objetos (POO) en Python, modelando la clase Seguro para centralizar la administración de servicios financieros en una estructura lógica. Mediante el uso de atributos (tipo, cobertura, costo) se establece la identidad y las condiciones comerciales de cada contrato, mientras que las funciones (leerdatos(), mostrardatos()) permiten la interacción directa con el usuario para la captura y visualización organizada de la información. Este enfoque facilita la creación de los objetos independientes seguro1 y seguro2, demostrando cómo la POO permite gestionar múltiples registros de forma modular, profesional y eficiente.


![texto del vínculo](https://raw.githubusercontent.com/RicardoLopezCalderon128/Act-25-clase-seguro-Lopez-Ricardo-0085./refs/heads/main/infografia%20Seguro.png)

```python
# --- CLASE SEGURO EN PYTHON ---

class Seguro:
    def __init__(self):
        # Inicializamos los atributos vacíos para luego llenarlos
        self.tipo = ""
        self.cobertura = ""
        self.costo = 0.0

    # Función para capturar los datos por teclado
    def leerdatos(self, numero_objeto):
        print(f"\n--- Ingresando datos para el Seguro {numero_objeto} ---")
        self.tipo = input("Ingrese el tipo de seguro (ej. Auto, Vida, Hogar): ")
        self.cobertura = input("Ingrese el nivel de cobertura (ej. Total, Básica): ")
        self.costo = float(input("Ingrese el costo mensual del seguro: "))

    # Función para mostrar la información del objeto
    def mostrardatos(self, nombre_objeto):
        print(f"\n📋 INFORMACIÓN DE {nombre_objeto.upper()}:")
        print(f"🔹 Tipo de Seguro: {self.tipo}")
        print(f"🔹 Cobertura: {self.cobertura}")
        print(f"🔹 Costo Mensual: ${self.costo:,.2f}")

# --- CREACIÓN DE OBJETOS E INSTANCIACIÓN ---

print("=== SISTEMA DE GESTIÓN DE SEGUROS EN PYTHON ===")

# 1. Creamos los 2 objetos (instancias de la clase Seguro)
seguro1 = Seguro()
seguro2 = Seguro()

# 2. Llamamos a la función leerdatos() para cada objeto
seguro1.leerdatos(1)
seguro2.leerdatos(2)

# 3. Llamamos a la función mostrardatos() para ver el resultado
print("\n" + "="*45)
print("▶️ RESUMEN DE SEGUROS REGISTRADOS")
print("="*45)

seguro1.mostrardatos("seguro1")
seguro2.mostrardatos("seguro2")
print("Programa realizado por Ricardo Lopez NL = 29")
print("\n================================================")
```
