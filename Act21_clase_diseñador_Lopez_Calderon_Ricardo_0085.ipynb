```python
# Definición de la clase
class Diseñador:
    def __init__(self, nombre, creatividad, detalle, organizacion):
        # Atributos de instancia
        self.nombre = nombre
        self.creatividad = creatividad
        self.detalle = detalle
        self.organizacion = organizacion

    # Función 1: Uso de software
    def uso_de_software(self, herramienta):
        return f"{self.nombre} está utilizando {herramienta} para el proyecto."

    # Función 2: Investigación y análisis
    def investigacion_y_analisis(self):
        return f"{self.nombre} está analizando las tendencias del mercado (Nivel de detalle: {self.detalle})."

    # Función 3: Desarrollo de soluciones
    def desarrollo_de_soluciones(self):
        return f"{self.nombre} ha desarrollado una solución creativa basada en su {self.creatividad}."

# --- Creación de Objetos (Instancias) ---

# diseñador1 con datos de ejemplo
diseñador1 = Diseñador("Alex", "Alta (Estilo Minimalista)", "90%", "Excelente")

# diseñador2 con datos de ejemplo
diseñador2 = Diseñador("Bárbara", "Vanguardista", "85%", "Media")

# --- Ejecución y Mensajes ---

print("--- DATOS DEL DISEÑADOR 1 ---")
print(f"Nombre: {diseñador1.nombre}")
print(diseñador1.uso_de_software("Adobe Illustrator"))
print(diseñador1.investigacion_y_analisis())
print(diseñador1.desarrollo_de_soluciones())

print("\n--- DATOS DEL DISEÑADOR 2 ---")
print(f"Nombre: {diseñador2.nombre}")
print(diseñador2.uso_de_software("Figma"))
print(diseñador2.investigacion_y_analisis())
print(diseñador2.desarrollo_de_soluciones())

print("fin del programa")
print("programa diseñado por Lopez Ricardo NL = 29")
```
