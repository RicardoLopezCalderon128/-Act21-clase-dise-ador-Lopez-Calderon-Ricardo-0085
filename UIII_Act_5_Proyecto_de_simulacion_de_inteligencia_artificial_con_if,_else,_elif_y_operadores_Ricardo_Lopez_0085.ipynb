* Para crear un programa en Python que simule una inteligencia artificial básica actuando como un traductor de español a inglés, utilizaremos la estructura de decisión múltiple if-elif-else y los operadores relacionales y lógicos
* Este código permite una interacción directa con el usuario para traducir palabras específicas, basándose en la lógica de selección de casos

* A continuación, se presenta el código siguiendo tus restricciones de evitar el uso de clases, funciones o ciclos de repetición:

```python
# ==========================================
# TRADUCTOR INTELIGENTE ESPAÑOL - INGLÉS
# Simulación de IA usando solamente:
# if, elif, else y operadores
# Sin funciones
# Sin ciclos
# Sin programación orientada a objetos
# ==========================================

print("===================================")
print("   IA TRADUCTORA ESPAÑOL-INGLÉS")
print("===================================")

nombre = input("Ingrese el nombre de la PersonaQueTraduceDeEspañolaIngles: ")
print("Hola", nombre)
print("Este sistema simula una inteligencia artificial traductora.")
print("")

print("Palabras disponibles para traducir:")
print("- hola")
print("- adios")
print("- perro")
print("- gato")
print("- comida")

print("")

palabra = input("Escriba una palabra en español: ")

# Condicionales principales
if palabra == "hola":
    print("IA:", nombre + ", la traducción es: hello")

elif palabra == "adios":
    print("IA:", nombre + ", la traducción es: goodbye")

elif palabra == "perro":
    print("IA:", nombre + ", la traducción es: dog")

elif palabra == "gato":
    print("IA:", nombre + ", la traducción es: cat")

elif palabra == "comida":
    print("IA:", nombre + ", la traducción es: food")



else:
    print("IA: Lo siento,", nombre + ", no conozco esa palabra.")
    print("Intente con otra palabra de la lista.")

print("")
print("===================================")
print("ANÁLISIS INTELIGENTE DE LA IA")
print("===================================")

# Uso de operadores lógicos
if palabra == "hola" or palabra == "adios":
    print("La palabra pertenece a SALUDOS.")

elif palabra == "perro" or palabra == "gato":
    print("La palabra pertenece a ANIMALES.")

elif palabra == "comida" or palabra == "agua":
    print("La palabra pertenece a ALIMENTOS.")

else:
    print("La IA no pudo clasificar la palabra.")

print("")
print("===================================")
print("      FIN DEL PROGRAMA    ")
print("===================================")

print("peograma realizado por lopez Ricardo NL = 33")
```
