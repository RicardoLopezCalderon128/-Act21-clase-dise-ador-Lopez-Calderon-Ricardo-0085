#### redaccion del problema
* Este programa simula un sistema de evaluación inteligente para determinar si una persona califica como aval o fiador para un crédito.
* Utiliza una estructura de control de flujo basada en decisiones lógicas y repetición, cumpliendo estrictamente con tus restricciones de no usar funciones, clases ni estructuras complejas

```python
# --- SIMULADOR DE IA: EVALUADOR DE AVAL O FIADOR ---

print("--- Sistema de Inteligencia Lógica de Crédito ---")
continuar = "si"

# El ciclo while permite que el programa sea interactivo y se repita [3, 4]
while continuar == "si" or continuar == "Si":
    print("\nIniciando nueva evaluación...")

    # Captura de datos iniciales mediante input [5, 6]
    nombre = input("Nombre del aspirante: ")
    edad = int(input("Ingrese la edad: "))

    # Estructura condicional simple para verificar la mayoría de edad [7, 8]
    if edad < 18:
        print("Resultado: RECHAZADO. El aval debe ser mayor de edad.")
    else:
        # Si es mayor de edad, se solicitan datos financieros
        print("Persona mayor de edad. Procediendo a análisis financiero...")

        ingresos = float(input("Ingresos mensuales líquidos: "))
        deudas = float(input("Total de deudas mensuales actuales: "))
        monto_a_garantizar = float(input("Monto de la cuota que desea garantizar: "))
        historial_bueno = input("¿Tiene buen historial crediticio? (si/no): ")

        # Uso de operadores aritméticos para calcular la capacidad de pago [9, 10]
        # Se asume que el aval no debe comprometer más del 30% de su ingreso disponible
        capacidad_disponible = (ingresos - deudas) * 0.30

        # Inteligencia de decisión utilizando if-elif-else y operadores lógicos [2, 11]
        if historial_bueno == "no":
            print(f"Resultado: RECHAZADO. {nombre} tiene historial negativo.")

        elif ingresos < (monto_a_garantizar * 3):
            # Operador relacional para comparar ingresos mínimos [12, 13]
            print(f"Resultado: RECHAZADO. Ingresos insuficientes para el monto solicitado.")

        elif capacidad_disponible >= monto_a_garantizar:
            # Combinación de condiciones para la aprobación final
            print(f"--- EVALUACIÓN EXITOSA ---")
            print(f"El aspirante {nombre} califica como Aval/Fiador.")
            print(f"Capacidad de respaldo mensual: ${capacidad_disponible}")

        else:
            # Caso por defecto si no se cumplen los requisitos de capacidad [2]
            print(f"Resultado: RECHAZADO. El riesgo financiero es muy alto para el nivel de ingresos.")

    # Control del ciclo para decidir si se evalúa a otra persona [14, 15]
    continuar = input("\n¿Desea realizar otra evaluación? (si/no): ")

print("--- Gracias por usar el Sistema Evaluador. Proceso finalizado ---")
```
