```python
# --- SIMULADOR DE IA: ASISTENTE LÓGICO PARA ELECTRICISTAS ---

# Inicializamos la variable de control para el menú interactivo [3]
opcion = "0"

# El ciclo 'while' mantiene el programa activo hasta que el usuario decida salir [4]
while opcion != "3":
    print("\n==========================================")
    print("   ⚡ IA: ASISTENTE DEL ELECTRICISTA ⚡   ")
    print("==========================================")
    print("1. Consultar herramienta: Alicate")
    print("2. Diagnóstico de Circuito Lógico")
    print("3. Salir del sistema")

    # Captura de la elección del usuario de forma interactiva [5]
    opcion = input("\nSeleccione una opción (1, 2 o 3): ")

    # Estructura condicional múltiple para manejar el menú [2, 6]
    if opcion == "1":
        print("\n--- MODO: EXPERTO EN ALICATES ---")
        print("Tipos disponibles: universal, corte, punta")
        tipo_alicate = input("¿Qué tipo de alicate tienes en la mano?: ")

        # Lógica de IA para identificar funciones específicas mediante comparaciones [7, 8]
        if tipo_alicate == "universal" or tipo_alicate == "Universal":
            print("IA dice: El alicate universal sirve para apretar, doblar y cortar cables.")
        elif tipo_alicate == "corte" or tipo_alicate == "Corte":
            print("IA dice: Este alicate es específico para seccionar conductores de cobre o aluminio.")
        elif tipo_alicate == "punta" or tipo_alicate == "Punta":
            print("IA dice: Ideal para sujetar piezas pequeñas en lugares de difícil acceso.")
        else:
            print("IA dice: Herramienta no reconocida en mi base de datos actual.")

    elif opcion == "2":
        print("\n--- MODO: DIAGNÓSTICO DE CIRCUITO ---")
        # El electricista ingresa datos numéricos del sistema [9]
        voltaje = float(input("Ingresa el voltaje medido (V): "))
        resistencia = float(input("Ingresa la resistencia del cable (Ohm): "))

        # Uso de operadores aritméticos para calcular la intensidad (Ley de Ohm) [10, 11]
        intensidad = voltaje / resistencia

        # Evaluación lógica utilizando operadores relacionales [7]
        if intensidad > 15.0:
            # Si la corriente supera los 15 Amperios, hay riesgo [5]
            print(f"¡ALERTA!: La corriente es de {intensidad} A. Es demasiado alta.")
            print("IA recomienda: Usa tu alicate de corte para desconectar y revisar el aislante.")
        elif intensidad > 0 and intensidad <= 15.0:
            print(f"Estado: Seguro. La corriente es de {intensidad} A.")
            print("IA recomienda: Puedes proceder con la instalación.")
        else:
            # Caso para valores negativos o cero [5]
            print("IA dice: No se detecta flujo de corriente. Revisa la fuente de poder.")

    elif opcion == "3":
        # Finalización del ciclo [12, 13]
        print("\nCerrando sistema de IA... ¡Seguridad ante todo, electricista!")

    else:
        # Manejo de opciones no válidas [14, 15]
        print("\nOpción inválida. Por favor, ingresa 1, 2 o 3.")

print("==========================================")
```
